```python
def label_to_np(label: str) -> np.array:
    isic_ar = 'MEL,NV,BCC,AK,BKL,DF,VASC,SCC'.split(',')
    mapping = {
        'AKIEC': 'AK'
    }
    zero_arr = np.zeros((len(isic_ar)))
    refined_label = label.upper()
    refined_label = refined_label if refined_label not in mapping else mapping[refined_label]
    
    if refined_label not in isic_ar:
        return zero_arr
    
    ind = isic_ar.index(refined_label)
    res = zero_arr
    res[ind] = 1.0

    return res

```

```python
label_to_np('akiec')
```

```python
converted_columns = df.apply(lambda x: label_to_np(x), axis='columns')
```
